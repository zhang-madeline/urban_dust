# AD-Net Analysis
Data is from AD-Net (https://www-lidar.nies.go.jp/AD-Net/): Asian dust and aerosol lidar observation network.

## Scripts Information

adnet_download.py: Downloads directly from website. Alter variables file_path and start_year/end_year. <br>
adnet_data_analysis.py: Analyzes station data availability. Alter file_path. <br>

## Raw file data variables
PPPYYMMDD.ncdf <br>
PPP:3 letter code <br>
('CHB', 'FKE', 'FKO', 'HED', 'JEJ', 'MTS', 'NGS', <br>
'NIG', 'OSK', 'PHM', 'SEO', 'SND', 'SNS', 'SPR', <br>
'TKB', 'TKO', 'TYM', 'ULN', 'ULS', 'UTY', 'ZMY') <br>
YYMMDD:date in UTC <br>
<br>
**Dimensions:** <br>
TIME: UTC Start time (sec from 00Z) # every 15 min, 96 prof./day <br>
HEIGHTA: altitude from ground (km) for attenuated backscatter and volume depol. #up to 18km <br>
HEIGHTE: altitude from ground (km) for extinction data #up to 9 km <br>
<br>
**Variables:** <br>
ABSC532(T*HA): 532 nm attenuated backscatter coefficient (/m/sr) <br>
ABSC1064(T*HA): 1064 nm (/m/sr) <br>
DEP532(T*HA): 532 nm volume depolarization ratio #S/P <br>
EXT532(T*HE): 532 nm aerosol extinction coefficient (/m) <br>
EXTD532(T*HE): 532 nm dust extinction coefficient (/m) <br>
EXTS532(T*HE): 532 nm spherical particle extinction coefficient (/m) <br>
ADEP532(T*HE): 532 nm aerosol depolarization ratio <br>
MXLHT(T): estimated mixing layer height (m) <br>
<br>
**Global attribute:** <br>
location: station name (city, country) <br>
latitude: latitude (degree) <br>
longitude: longitude(degree) <br>
altitude: station altitude above mean sea level: (m) <br>
date: date of observation in UTC <br>
lidar ratio(sr): S1 <br>
molecular backscatter(per msr):532 nm molecular (Rayleigh) backscatter coefficient (/m/sr), derived from CIRA86 <br>
#following are mainly for records in NIES <br>
analysis time: UNIX time of the analysis <br>
dep ratio: ratio of sensitivity of detectors for perp/para <br>
ir vis raito: ratio of sensitivity of detectors for 1064/532 <br>
int abc ratio: ratio of abc532 to intensity <br>
now: last day of analysis period <br>
number_of_days: days utilized for statistical analysis <br>
